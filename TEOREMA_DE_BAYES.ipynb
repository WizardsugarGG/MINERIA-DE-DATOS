```python
from re import X
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

#sentiment labelled senteces

df= pd.read_csv("amazon_cells_labelled.txt",
               sep='	',
               names=['review','sentiment'])
print(df.head())

#contamos el numero  de palabras en cada texto
print(df['sentiment'].value_counts())

#crear el modelo de naive bayes multinomial
vectorizer= CountVectorizer()
X= vectorizer.fit_transform(df['review'])
y= df['sentiment']

modelo = MultinomialNB()
modelo.fit(X,y)

#prediccio
ejemplos = ["I LOVE THIS MOVIE","THIS IS THE WORST MOVIE"]
X_ejemplos = vectorizer.transform(ejemplos)
pred = modelo.predict(X_ejemplos)
print(pred)
#x texto y variable


#.fit es para entrenar el modelo
```

```python
#evaluar modelo

X_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.3)#divimos los datos e entremaniento y prueba

modelo.fit(X_train, y_train)
y_pred = modelo.predict(x_test)

accuracy = accuracy_score(y_test, y_pred)
print(f"precision:{accuracy: .2f}")
```
