```python
# 1. Crear un array de 10 números aleatorios enteros entre 0 y 100
array1 = np.random.randint(0, 101, 10)
print("Array 10 enteros:", array1)
```

```python
# 2. Crear un array de 5 números aleatorios decimales entre 0 y 1
array2 = np.random.rand(5)
print("Array 5 decimales:", array2)
```

```python
# 3. Crear dos arrays de números aleatorios enteros de longitud 5 y concatenarlos
a = np.random.randint(0, 100, 5)
b = np.random.randint(0, 100, 5)
concatenado = np.concatenate((a, b))
print("Array A:", a)
print("Array B:", b)
print("Concatenado:", concatenado)
```

```python
# 4. Crear un array de 10 números aleatorios enteros y separarlo en dos arrays de 5
array3 = np.random.randint(0, 100, 10)
parte1, parte2 = np.split(array3, 2)
print("Array original:", array3)
print("Parte 1:", parte1)
print("Parte 2:", parte2)
```

```python
# 5. Crear una matriz 3x3 con números aleatorios decimales entre 0 y 1
matriz = np.random.rand(3, 3)
print("Matriz 3x3:\n", matriz)
```

```python
# 6. Crear un array de 10 números aleatorios enteros y seleccionar 3 al azar
array4 = np.random.randint(0, 100, 10)
seleccion = np.random.choice(array4, 3, replace=False)
print("Array:", array4)
print("3 elementos al azar:", seleccion)
```

```python
# 7. Crear un array de 10 números aleatorios enteros entre 0 y 100 y calcular la media
array5 = np.random.randint(0, 101, 10)
media = np.mean(array5)
print("Array:", array5)
print("Media:", media)
```
